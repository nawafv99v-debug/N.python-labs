```python
import os
import subprocess

class Person:
    def __init__(self, name, id):
        self.name = name
        self.__id = id

    def get_id(self):
        return self.__id

class Student(Person):
    def __init__(self, name, id, grade):
        super().__init__(name, id)
        self.grade = grade

    def display(self):
        print(f"Name: {self.name}, ID: {self.get_id()}, Grade: {self.grade}")

def load_students():
    students = []
    path = "Nmyfil.txt"
    if os.path.exists(path):
        with open(path, "r") as f:
            for line in f:
                name, id, grade = line.strip().split(",")
                students.append(Student(name, id, grade))
    return students

def save_students(students):
    path = "Nmyfil.txt"
    with open(path, "w") as f:
        for s in students:
            f.write(f"{s.name},{s.get_id()},{s.grade}\n")
    folder = os.path.abspath(os.path.dirname(path))
    print("File saved at:", os.path.abspath(path))
    subprocess.Popen(f'explorer "{folder}"')

students = load_students()

while True:
    print("\n1. Add Student")
    print("2. View Students")
    print("3. Update Student")
    print("4. Delete Student")
    print("5. Exit")

    choice = input("Choose: ")

    if choice == "1":
        name = input("Enter name: ")
        id = input("Enter ID: ")
        grade = input("Enter grade: ")
        if not grade.isdigit():
            print("Grade must be a number!")
            continue
        students.append(Student(name, id, grade))
        save_students(students)

    elif choice == "2":
        students = load_students()
        if students:
            for s in students:
                s.display()
        else:
            print("No students found!")

    elif choice == "3":
        id = input("Enter ID to update: ")
        students = load_students()
        found = False
        for s in students:
            if s.get_id() == id:
                s.name = input("New name: ")
                s.grade = input("New grade: ")
                found = True
                break
        if not found:
            print("Student ID not found!")
        save_students(students)

    elif choice == "4":
        id = input("Enter ID to delete: ")
        students = load_students()
        new_students = [s for s in students if s.get_id() != id]
        if len(new_students) == len(students):
            print("Student ID not found!")
        else:
            students = new_students
            save_students(students)

    elif choice == "5":
        break

    else:
        print("Invalid choice!")
```
