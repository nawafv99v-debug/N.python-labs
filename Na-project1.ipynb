```python
import os

file_path = r"C:\Users\nawaf\OneDrive\Desktop"

class Person:
    def __init__(self, name, id):
        self.name = name
        self.__id = id

    def get_id(self):
        return self.__id


class Student(Person):
    def __init__(self, name, id, grade):
        super().__init__(name, id)
        self.grade = grade

    def display(self):
        print(f"Name: {self.name}, ID: {self.get_id()}, Grade: {self.grade}")


def nawaf_students():
    students = []
    if os.path.exists(file_path):
        with open(file_path, "r") as f:
            for line in f:
                try:
                    name, id, grade = line.strip().split(",")
                    students.append(Student(name, id, grade))
                except:
                    pass
    return students


def save_students(students):
    with open(file_path, "w") as f:
        for s in students:
            f.write(f"{s.name},{s.get_id()},{s.grade}\n")


students = nawaf_students()

while True:
    print("\n1. Add Student")
    print("2. View Students")
    print("3. Update Student")
    print("4. Delete Student")
    print("5. Exit")

    choice = input("Choose: ")

    if choice == "1":
        name = input("Enter name: ")
        id = input("Enter ID: ")
        grade = input("Enter grade: ")

        if not grade.isdigit():
            print("Grade must be a number!")
            continue

        students.append(Student(name, id, grade))
        save_students(students)

    elif choice == "2":
        for s in students:
            s.display()

    elif choice == "3":
        id = input("Enter ID to update: ")
        for s in students:
            if s.get_id() == id:
                s.name = input("New name: ")
                s.grade = input("New grade: ")
                break
        save_students(students)

    elif choice == "4":
        id = input("Enter ID to delete: ")
        students = [s for s in students if s.get_id() != id]
        save_students(students)

    elif choice == "5":
        break

    else:
        print("Invalid choice!")
```
