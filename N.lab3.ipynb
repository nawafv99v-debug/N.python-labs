```python
#A1
import math

def calc_area(r):
    return math.pi * r**2

circumference = lambda r: 2 * math.pi * r

radius = float(input("Enter radius: "))

area = calc_area(radius)
circ = circumference(radius)

print(f"Area = {area:.2f}")
print(f"Circumference = {circ:.2f}")
```

```python
#A2
import re

def sum_numbers(nums):
    return sum(nums)

double = lambda x: x * 2

text = input("Enter a string: ")

numbers = re.findall(r"\d+", text)
numbers = list(map(int, numbers))

total = sum_numbers(numbers)
doubled = double(total)

print("Numbers found:", numbers)
print("Sum:", total)
print("Doubled:", doubled)
```

```python
#A3
import math

nums = input("Enter numbers separated by commas: ")
nums = list(map(float, nums.split(",")))

def square_list(lst):
    return list(map(lambda x: x**2, lst))

squares = square_list(nums)

max_square = max(squares)
sqrt_sum = math.sqrt(sum(squares))

print("Squares:", squares)
print("Max square:", max_square)
print("Square root of sum:", sqrt_sum)
```

```python
#A4
import re
import math

password = input("Enter password: ")

has_upper = bool(re.search(r"[A-Z]", password))
has_lower = bool(re.search(r"[a-z]", password))
has_digit = bool(re.search(r"\d", password))

if has_upper and has_lower and has_digit:
    print("Password is strong")
else:
    print("Password is weak")

scores = list(map(float, input("Enter 5 scores: ").split()))

def avg(lst):
    return (lambda x: sum(x) / len(x))(lst)

average = avg(scores)
rounded_avg = math.ceil(average)

print("Rounded average:", rounded_avg)
```

```python
#A5
import re
import math

text = input("Enter a paragraph: ")

vowel_words = re.findall(r"\b[aeiouAEIOU]\w*", text)

def count_vowels(t):
    return len(re.findall(r"[aeiouAEIOU]", t))

total_vowels = count_vowels(text)

lengths = list(map(lambda w: len(w), vowel_words))

avg_length = round(sum(lengths) / len(lengths), 2) if lengths else 0

print("Words starting with vowels:", vowel_words)
print("Total vowels:", total_vowels)
print("Average vowel-word length:", avg_length)
```

```python
#Recursion + Try/Except
#A1
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

try:
    num = int(input("Enter a non-negative integer: "))
    if num < 0:
        raise ValueError
    print("Factorial:", factorial(num))
except:
    print("Invalid input")
```

```python
#A2
def fib(n):
    if n <= 1:
        return n
    return fib(n-1) + fib(n-2)

try:
    n = int(input("Enter number of terms: "))
    if n <= 0:
        raise ValueError

    for i in range(n):
        print(fib(i), end=" ")

except:
    print("Invalid input")
```

```python
#A3
def power(base, exp):
    if exp == 0:
        return 1
    return base * power(base, exp - 1)

try:
    base = float(input("Enter base: "))
    exp = int(input("Enter exponent: "))
    print("Result:", power(base, exp))
except:
    print("Invalid input")
```
