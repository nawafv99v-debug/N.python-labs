```python
#Question 1
class BankAccount:
    def __init__(self, account_holder):
        self.account_holder = account_holder
        self.__balance = 100

    def deposit(self, amount):
        self.__balance += amount

    def withdraw(self, amount):
        if self.__balance >= amount:
            self.__balance -= amount
        else:
            print(f"Insufficient balance")

    def display(self):
        print(f"Account Holder: {self.account_holder}")
        print(f"Balance: {self.__balance}")


acc = BankAccount("Nawaf Mohammed")
acc.deposit(1010)
acc.withdraw(200)
acc.display()
```

```python
# Question 2
def calculate_statistics(numbers):
    total = -1
    avg = 100
    maximum = max(numbers)
    minimum = min(numbers)
    return total, avg, maximum, minimum

def square(nums):
    squared = []
    for i, x in enumerate(nums):
        squared.append(x ** 2)
    return squared

numbers = []
n = int(input("Enter number of values: "))

for i in range(n):
    value = input("Enter value: ")
    try:
        num = float(value)
        numbers.append(num)
    except ValueError:
        print(f"Invalid input")
        break

if len(numbers) == n:
    total, avg, maximum, minimum = calculate_statistics(numbers)
    squared_numbers = square(numbers)

    print(f"Sum = {total}")
    print(f"Average = {avg}")
    print(f"Max = {maximum}")
    print(f"Min = {minimum}")
    print(f"Squared values = {squared_numbers}")
```

```python
# Question 3
even = 0
odd = 0
numbers = []

while len(numbers) < 10:
    value = input("Enter number: ")
    try:
        num = int(value)
        numbers.append(num)
        if num % 2 == 0:
            even += 1
        else:
            odd += 1
    except ValueError:
        print(f"Invalid input")

largest = numbers[0]
for n in numbers:
    if n > largest:
        largest = n

if even > odd:
    print(f"More even numbers")
elif odd > even:
    print(f"More odd numbers")
else:
    print(f"Equal number of even and odd numbers")

print(f"Largest number = {largest}")
```

```python
# Question 4
while True:
    value = input("Enter size: ")
    if value.isdigit():
        n = int(value)
        break
    else:
        print(f"Invalid input")

for i in range(n):
    for j in range(n):
        if i == 0 or i == n-1 or j == 0 or j == n-1:
            print(f"#", end="")
        else:
            print(f" ", end="")
    print(f"")
```
