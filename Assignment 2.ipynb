```python
#Q4
import cv2

img = cv2.imread(r"C:\Users\nawaf\OneDrive\Desktop\Faces.jpg")

cv2.rectangle(img, (50, 50), (200, 200), (0, 255, 0), 2)

cv2.circle(img, (300, 150), 50, (255, 0, 0), 2)

cv2.line(img, (0, 0), (400, 400), (0, 0, 255), 2)

cv2.imshow("Shapes", img)

cv2.waitKey(0)

cv2.destroyAllWindows()
```

```python
#Q5
import cv2  
img = cv2.imread(r"C:\Users\nawaf\OneDrive\Desktop\Faces.jpg") 
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)  
edges = cv2.Canny(gray, 100, 200)  
cv2.imshow("Edges", edges) 
cv2.waitKey(0) 
cv2.destroyAllWindows()
```

```python
import cv2

cap = cv2.VideoCapture(r"C:\Users\nawaf\OneDrive\Desktop\arabsstock_P412435.mp4")

while True:
    ret, frame = cap.read()

    if not ret:
        print("Video ended or cannot open")
        break

    cv2.imshow("Video", frame)

    if cv2.waitKey(25) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
