```python
try:
    num1 = float(input("Enter the first number: "))
    num2 = float(input("Enter the second number: "))

    result = num1 / num2

except ZeroDivisionError:
    print("Error: Cannot divide by zero!")

except ValueError:
    print("Error: Please enter valid numbers, not text.")

else:
    print(f"The result of division is: {result}")
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self._age = age

    @property
    def age(self):
        return self._age

    @age.setter
    def age(self, new_age):
        if new_age > 0:
            self._age = new_age
        else:
            print("Invalid age!")

    def display(self):
        print(f"Name: {self.name}, Age: {self._age}")
```

```python
person1 = Person("NAWAF", 21) 
person1.display() 
person1.age = 22 
person1.display()
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary


class Manager(Employee):
    def add_bonus(self, bonus):
        self.salary += bonus

    def display(self):
        print(f"Manager Name: {self.name}")
        print(f"Total Salary (with bonus if any): {self.salary}")
```

```python
m1 = Manager("Nawaf", 4000)
m1.add_bonus(1000)
m1.display()
```

```python
class Student:
    def __init__(self):
        self.name = input("Enter your name: ")

        try:
            self.age = int(input("Enter your age: "))
            if self.age < 0:
                raise ValueError
        except ValueError:
            print("Invalid input")
            self.age = 0   # قيمة افتراضية حتى لا يحصل خطأ

    def display(self):
        print("name =", self.name)
        print("age =", self.age)
```

```python
student1 = Student()
student1.display()
```
